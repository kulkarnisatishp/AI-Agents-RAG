# Medium Exercise: String Reversal
Task: Write a function named reverse_string that takes a string as input and returns it reversed.

```python
def reverse_string(input_string):
  reverse_string = input_string[::-1]
  return reverse_string
reverse_string("Diogo")
```

# Medium Exercise: Anagram Finder
Task: Write a function named are_anagrams that takes two strings and returns True if they are anagrams, False otherwise.

```python
string_example = "Listen "
string_example_cleaned = string_example.replace(" ", "").lower()
string_example_cleaned_sorted = sorted(string_example_cleaned)
print(string_example_cleaned_sorted)
```

```python
string_example = "silent "
string_example_cleaned = string_example.replace(" ", "").lower()
string_example_cleaned_sorted = sorted(string_example_cleaned)
print(string_example_cleaned_sorted)
```

```python
def are_anagrams(string1, string2):
  string1_cleaned_sorted = sorted(string1.replace(" ", "").lower())
  string2_cleaned_sorted = sorted(string2.replace(" ", "").lower())
  return string1_cleaned_sorted == string2_cleaned_sorted
are_anagrams("silent", "listen")
```
