#Easy Exercise: Calculate the Sum of a List

Task: Write a function named calculate_sum that takes a list of numbers as input and returns their sum.

```python
def calculate_sum(numbers):
  return sum(numbers)

initial_list = [1,2,3,4]
print(calculate_sum(initial_list))
```

# Easy Exercise: Area of a Rectangle
Task: Write a function named calculate_area that takes two parameters (length and width) and returns the area.

```python
def calculate_area(length, width):
  area = length * width
  return area

calculate_area(5,4)
```
