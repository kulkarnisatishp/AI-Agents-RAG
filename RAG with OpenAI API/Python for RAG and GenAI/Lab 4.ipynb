# Easy Exercise: Create a Simple Class for a Book
**Description**: Represent a book using a class.

Task 1: Define the Class and Attributes
* Create a class Book with attributes: title, author, and pages.
* Test by creating a book object and checking its attributes.


Task 2: Add a Method
* Add book_info() to return: "Title: [title], Author: [author], Pages: [pages]".
* Test by calling the method on your book object.


```python
class Book:
  def __init__(self, title, author, pages):
    self.title = title
    self.author = author
    self.pages = pages

  def book_info(self):
    return f"Title: {self.title}, author: {self.author}, pages: {self.pages}"

my_book = Book("Harry Potter", "J. K. Rowling", 200)
print(my_book.title)
print(my_book.author)
print(my_book.pages)

my_book.book_info()
```
