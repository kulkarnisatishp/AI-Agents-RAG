# Easy Exercise: Create a Simple Class for a Rectangle
**Description**: Represent a rectangle using a class.

Task 1: Define the Class and Attributes
* Create Rectangle with attributes length and width.
* Test by creating an object and printing its attributes.


Task 2: Add the area Method
* Add area() that returns length * width.
* Test by calling .area() on your object.


Task 3: Add the perimeter Method
* Add perimeter() that returns 2 * (length + width).
* Test by calling .perimeter() on your object.
