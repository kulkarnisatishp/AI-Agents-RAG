# Medium Exercise: Implement a Simple Banking System
**Description**: Create a banking system with deposits, withdrawals, and balance checks.

Task 1: Define the Class and Attributes
* Create BankAccount with attributes account_holder and balance (default 0.0).
* Test by creating an account and printing the attributes.


Task 2: Add the deposit Method
* Add deposit(amount) that increases balance and prints confirmation.


Task 3: Add the withdraw Method
* Add withdraw(amount) that deducts balance if funds are sufficient, otherwise prints "Insufficient funds".


Task 4: Add the get_balance Method
* Add get_balance() that returns the current balance.
* Test by printing the result after deposits and withdrawals.
