```python
concepts = [
    "Programmer excitedly typing, coffee in hand, yelling 'Draw me a rainbow-breathing llama!'",
    "Guard dog wearing sunglasses labeled 'API KEY' blocking the door",
    "Tree branch holding a bottle saying 'I’m citrus-powered!'",
    "User screaming 'It actually worked!' while the monitor glows",
    "Photocopier labeled 'GPT-Image-1' spitting out a perfect clone",
    "AI wearing a beret, copying artwork with a magnifying glass",
    "Each generation slightly blurrier: classic 'copy of a copy' VHS gag",
    "GIF icon under disco lights with caption 'Infinite loop, infinite fun'",
    "Robot cashier ringing up an invoice: 'That’ll be 0.19 cents, pal'",
    "Thrilled wallet with sunglasses saying 'Still got it!'",
    "Bookworm robot juggling reference links",
    "Triumphant coder holding a glowing paintbrush, caption 'Go forth and generate!'"
]
```

# Setup

```python
import base64, os, imageio
from openai import OpenAI
from PIL import Image
```

```python
%cd /content/drive/MyDrive/GenAI/OpenAI/Image Generation
```

```python
# Set OpenAI API key in the environment
from google.colab import userdata
openai_api = userdata.get('genai_course')
os.environ['OPENAI_API_KEY'] = openai_api
client = OpenAI()
```

# Meme generator

```python
for idx, prompt_text in enumerate(concepts, start=1):
    # Generate the image for the current concept
    img_resp = client.images.generate(
        model="gpt-image-1",
        prompt=prompt_text,
        n=1,
        size="1024x1024"
    )

    # Decode Base64 image bytes
    img_bytes = base64.b64decode(img_resp.data[0].b64_json)

    # Build a safe filename (e.g., concept_01.png, concept_02.png, ...)
    filename = f"concept_{idx:02d}.png"

    # Write the image to the current working directory
    with open(filename, "wb") as f:
        f.write(img_bytes)

    # Optional: print confirmation for tracking in Colab logs
    print(f"Saved {filename} for prompt: {prompt_text}")
```
