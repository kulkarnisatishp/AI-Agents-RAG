# Require the approval

```python
# Call the OpenAI API with the responses endpoint
response2 = client.responses.create(
    model = "gpt-5.6-luna",
    input="""
    Use Zapier Gmail.

    Find my most recent Udemy email.

    Use the Gmail find_email action:
    {
      "query": "Udemy",
      "max_results": 1,
      "paging_token": "FIRST_PAGE"
    }

    Return:
    subject, sender, date, and snippet.
    """,
    tools = [
        {
            "type": "mcp",
            "server_label": "zapier",
            "server_url": f"https://mcp.zapier.com/api/v1/connect?token={ZAPIER_TOKEN}",
            "require_approval": "always"
        }
    ]
)
```

```python
# Explore the output with the approval
response2.output[1].id
```

```python
# Include the approval in the following conversation
response3 = client.responses.create(
    model="gpt-5.6-luna",
    tools=[
        {
            "type": "mcp",
            "server_label": "zapier",
            "server_url": f"https://mcp.zapier.com/api/v1/connect?token={ZAPIER_TOKEN}",
        }
    ],
    input=[
        {
            "type": "mcp_approval_response",
            "approve": True,
            "approval_request_id": response2.output[1].id
        }
    ],
    previous_response_id=response2.id
)

print(response3.output_text)
```

# Authentification with Stripe

```python
# Get the stripe key
from google.colab import userdata
stripe_key = userdata.get('STRIPE_KEY')
```

```python
# Call the API with Stripe
response6 = client.responses.create(
    model = "gpt-5.6-luna",
    input = "Create a customer called Andrei with email andrei@gmail.com",
    tools = [
        {
            "type": "mcp",
            "server_label": "stripe",
            "server_url": "https://mcp.stripe.com",
            "headers": {
                "Authorization": f"Bearer {stripe_key}"
            }
        }
    ]
)
Markdown(response6.output_text)
```

```python
# Explore the output
response6.output[1]
```

```python
# Create a flow for approval
def stripe_request(query):
  request = client.responses.create(
      model = "gpt-5.6-luna",
      input = query,
      tools = [
          {
              "type": "mcp",
              "server_label": "stripe",
              "server_url": "https://mcp.stripe.com",
              "headers": {
                  "Authorization": f"Bearer {stripe_key}"
              }
          }
      ]
  )
  print(request.output[1])

  # Ask the user if the request is accepted or not
  approval = input("Do you want to approve the request? (y/n)")
  if approval == "y":
    # Approve the request
    approval_request = client.responses.create(
        model = "gpt-5.6-luna",
        tools = [
            {
                "type": "mcp",
                "server_label": "stripe",
                "server_url": "https://mcp.stripe.com",
                "headers": {
                    "Authorization": f"Bearer {stripe_key}"
                }
            }
        ], # Added comma here
        previous_response_id=request.id,
        input = [
            {
                "type": "mcp_approval_response",
                "approve": True,
                "approval_request_id": request.output[1].id
            }
        ]
    )
    return Markdown(approval_request)
```

```python
# Apply the function
approval_request = stripe_request("Create a customer called Andrei with email andrei@gmail.com")
```
